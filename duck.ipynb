```python
import duckdb as dk

c = dk.sql("select *, UNNEST(resposta.results) as resultado from dados.jsonl")

c
```

```python


consult = dk.sql("""SELECT
    resultado.location.latitude as Latitude,
    resultado.location.longitude as Longitude,
    resultado.postalAddress.postalcode as CEP,
    resultado.postalAddress.locality as Cidade,
    resultado.postalAddress.sublocality as Bairro,
    resultado.postalAddress.addressLines[1] AS endereco,
    resultado.placeId,
    resultado.formattedAddress AS endereco_formatado,
    resultado.granularity as Precisão

FROM (
    SELECT
        UNNEST(resposta.results) AS resultado
    FROM dados.jsonl
);""")

consult.write_csv('resultado.csv')
```
